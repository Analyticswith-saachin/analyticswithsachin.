```python
import pandas as pd
import openpyxl
from openpyxl.chart import BarChart, Reference

# Step 1: Create main (empty) workout sheet with days pre-filled
data = {
    "Day": ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"],
    "Workout": ["" for _ in range(7)],
    "Weight (kg)": ["" for _ in range(7)],
    "Reps": ["" for _ in range(7)],
    "Total Volume (kg × reps)": ["" for _ in range(7)]
}

df = pd.DataFrame(data)

# Step 2: Create Excel file and write the workout log
file_name = "Daily_Workout_Dashboard.xlsx"
with pd.ExcelWriter(file_name, engine='openpyxl') as writer:
    df.to_excel(writer, sheet_name="Workout Log", index=False)

# Step 3: Load workbook and add summary
wb = openpyxl.load_workbook(file_name)
ws = wb["Workout Log"]

summary = wb.create_sheet("Summary")

summary["A1"] = "Metric"
summary["B1"] = "Value"

summary["A2"] = "Total Volume (Week)"
summary["B2"] = "=SUM('Workout Log'!E2:E8)"

summary["A3"] = "Average Weight (kg)"
summary["B3"] = "=AVERAGE('Workout Log'!C2:C8)"

summary["A4"] = "Average Reps"
summary["B4"] = "=AVERAGE('Workout Log'!D2:D8)"

# Step 4: Add bar chart for total volume per day
chart = BarChart()
chart.title = "Daily Total Volume"
chart.y_axis.title = "Volume (kg)"
chart.x_axis.title = "Day"

data = Reference(ws, min_col=5, min_row=1, max_row=8)  # E column (Total Volume)
cats = Reference(ws, min_col=1, min_row=2, max_row=8)  # A column (Days)
chart.add_data(data, titles_from_data=True)
chart.set_categories(cats)

ws.add_chart(chart, "G2")

# Step 5: Save the file
wb.save(file_name)

print(f"✅ Excel workout dashboard created: {file_name}")
```
